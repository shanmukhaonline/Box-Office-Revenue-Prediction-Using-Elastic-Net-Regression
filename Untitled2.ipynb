```python
def edit_distance(a, b):
    a, b = a.lower(), b.lower()

    previous = list(range(len(b) + 1))

    for i, char_a in enumerate(a, start=1):
        current = [i]

        for j, char_b in enumerate(b, start=1):
            current.append(min(
                current[-1] + 1,              # Insertion
                previous[j] + 1,               # Deletion
                previous[j-1] + (char_a != char_b)
            ))

        previous = current

    return previous[-1]


def correct_movie_title(query, limit=10):
    if not query.strip():
        print("Please enter a movie title.")
        return pd.DataFrame()

    results = [
        (title, edit_distance(query, title))
        for title in df["title"]
    ]

    results.sort(key=lambda x: (x[1], x[0]))

    output = pd.DataFrame(
        results[:limit],
        columns=["Suggested Movie", "Edit Distance"]
    )

    display(output)
    return output


correct_movie_title("Gladiatr")
```

```python
def build_suffix_array(text):
    return sorted(range(len(text)), key=lambda i: text[i:])


def suffix_array_search(query):
    query = query.lower().strip()

    if not query:
        return df.iloc[0:0]

    matches = []

    for title in df["title"]:
        text = title.lower()
        suffix_array = build_suffix_array(text)

        for pos in suffix_array:
            if text[pos:].startswith(query):
                matches.append(title)
                break

    result = df[df["title"].isin(matches)]

    print("Suffix Array results:", len(result))
    display(result[
        ["title", "year", "genres", "worldwide_revenue"]
    ])

    return result


suffix_array_search("mission")
```

```python
movie_graph = defaultdict(set)
genre_to_movies = defaultdict(list)

for idx, row in df.iterrows():
    for genre in row["genres"].split(","):
        genre = genre.strip().lower()

        if genre and genre != "unknown":
            genre_to_movies[genre].append(idx)

for genre, indices in genre_to_movies.items():
    for position, idx in enumerate(indices):
        start = max(0, position - 10)
        end = min(len(indices), position + 11)

        for neighbor in indices[start:end]:
            if neighbor != idx:
                movie_graph[idx].add(neighbor)
                movie_graph[neighbor].add(idx)


def find_movie_index(title):
    found = df.index[
        df["title"].str.casefold() == title.strip().casefold()
    ].tolist()

    return found[0] if found else None


def bfs_movie(title, limit=15):
    start = find_movie_index(title)

    if start is None:
        print("Movie not found. Check the exact title.")
        return []

    visited = {start}
    queue = deque([start])
    order = []

    while queue and len(order) < limit:
        node = queue.popleft()
        order.append(node)

        for neighbor in sorted(movie_graph[node]):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    print("BFS traversal")
    display(df.loc[order, ["title", "genres", "year"]])
    return order


def dfs_movie(title, limit=15):
    start = find_movie_index(title)

    if start is None:
        print("Movie not found. Check the exact title.")
        return []

    visited = set()
    stack = [start]
    order = []

    while stack and len(order) < limit:
        node = stack.pop()

        if node in visited:
            continue

        visited.add(node)
        order.append(node)

        for neighbor in sorted(movie_graph[node], reverse=True):
            if neighbor not in visited:
                stack.append(neighbor)

    print("DFS traversal")
    display(df.loc[order, ["title", "genres", "year"]])
    return order


# Test using the first movie in your uploaded sheet
example_title = df.iloc[0]["title"]
print("Starting movie:", example_title)

bfs_movie(example_title)
dfs_movie(example_title)
```

```python
def ford_fulkerson(capacity, source, sink):
    residual = [row[:] for row in capacity]
    n = len(residual)
    max_flow = 0

    def find_path(u, visited, flow):
        if u == sink:
            return flow

        visited[u] = True

        for v in range(n):
            if not visited[v] and residual[u][v] > 0:
                pushed = find_path(
                    v, visited,
                    min(flow, residual[u][v])
                )

                if pushed:
                    residual[u][v] -= pushed
                    residual[v][u] += pushed
                    return pushed

        return 0

    while True:
        pushed = find_path(source, [False] * n, float("inf"))

        if pushed == 0:
            break

        max_flow += pushed

    return max_flow


capacity = [
    [0, 10, 5, 0, 0],
    [0, 0, 4, 8, 0],
    [0, 0, 0, 3, 7],
    [0, 0, 0, 0, 10],
    [0, 0, 0, 0, 0]
]

print("Ford-Fulkerson maximum flow:",
      ford_fulkerson(capacity, 0, 4))
```

```python
def edmonds_karp(capacity, source, sink):
    residual = [row[:] for row in capacity]
    n = len(residual)
    max_flow = 0

    while True:
        parent = [-1] * n
        parent[source] = source
        queue = deque([source])

        while queue and parent[sink] == -1:
            u = queue.popleft()

            for v in range(n):
                if parent[v] == -1 and residual[u][v] > 0:
                    parent[v] = u
                    queue.append(v)

                    if v == sink:
                        break

        if parent[sink] == -1:
            break

        path_flow = float("inf")
        v = sink

        while v != source:
            u = parent[v]
            path_flow = min(path_flow, residual[u][v])
            v = u

        v = sink

        while v != source:
            u = parent[v]
            residual[u][v] -= path_flow
            residual[v][u] += path_flow
            v = u

        max_flow += path_flow

    return max_flow


print("Edmonds-Karp maximum flow:",
      edmonds_karp(capacity, 0, 4))
```

```python
def recommend_movies(genre=None, top_n=10):
    data = df.copy()

    data["rating"] = pd.to_numeric(data["rating"], errors="coerce")
    data["vote_count"] = pd.to_numeric(
        data["vote_count"], errors="coerce"
    ).fillna(0)

    if genre and genre.strip():
        data = data[
            data["genres"].str.contains(
                re.escape(genre.strip()), case=False, na=False
            )
        ]

    if data.empty:
        print("No movies found for that genre.")
        return data

    global_mean = df["rating"].mean()
    if pd.isna(global_mean):
        global_mean = 0

    # Bayesian-style weighted rating
    minimum_votes = max(
        1, int(data["vote_count"].quantile(0.75))
    )

    data["recommendation_score"] = (
        data["vote_count"] /
        (data["vote_count"] + minimum_votes)
    ) * data["rating"].fillna(global_mean) + (
        minimum_votes /
        (data["vote_count"] + minimum_votes)
    ) * global_mean

    result = data.sort_values(
        "recommendation_score", ascending=False
    ).head(top_n)

    display(result[
        ["title", "year", "genres", "rating",
         "vote_count", "recommendation_score",
         "worldwide_revenue"]
    ])

    return result


recommend_movies("Action", 10)
```

```python
def randomized_movie_sampling(sample_size=10):
    sample_size = min(max(0, int(sample_size)), len(df))

    result = df.sample(
        n=sample_size,
        replace=False,
        random_state=None
    )

    print("Randomly selected movies:", len(result))

    display(result[
        ["title", "year", "genres", "rating", "worldwide_revenue"]
    ])

    return result


randomized_movie_sampling(10)
```

```python
numeric_features = ["year", "rating", "vote_count"]
categorical_features = ["genres", "language", "countries"]

X = df[numeric_features + categorical_features].copy()
y = df["worldwide_revenue"].copy()

# A log transform reduces the effect of very high-grossing movies.
y_log = np.log1p(y)

X_train, X_test, y_train, y_test = train_test_split(
    X, y_log, test_size=0.20, random_state=42
)

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="constant", fill_value="Unknown"
    )),
    ("encoder", OneHotEncoder(
        handle_unknown="ignore", min_frequency=2
    ))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", ElasticNet(
        alpha=0.01,
        l1_ratio=0.5,
        max_iter=20000
    ))
])

model.fit(X_train, y_train)

predicted_log = model.predict(X_test)
actual_revenue = np.expm1(y_test)
predicted_revenue = np.maximum(0, np.expm1(predicted_log))

mae = mean_absolute_error(actual_revenue, predicted_revenue)
rmse = np.sqrt(mean_squared_error(actual_revenue, predicted_revenue))
r2 = r2_score(actual_revenue, predicted_revenue)

print("ELASTIC NET MODEL PERFORMANCE")
print("=" * 45)
print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print(f"MAE: ${mae:,.2f}")
print(f"RMSE: ${rmse:,.2f}")
print(f"R² score: {r2:.4f}")

comparison = pd.DataFrame({
    "Actual Revenue": actual_revenue,
    "Predicted Revenue": predicted_revenue
})

display(comparison.head(15))


def predict_movie_revenue(
    year, rating, vote_count, genres, language, countries
):
    new_movie = pd.DataFrame([{
        "year": year,
        "rating": rating,
        "vote_count": vote_count,
        "genres": genres,
        "language": language,
        "countries": countries
    }])

    prediction = max(0, np.expm1(model.predict(new_movie)[0]))

    print(f"Estimated worldwide revenue: ${prediction:,.2f}")
    return prediction


# Example prediction
predict_movie_revenue(
    2026, 8.0, 10000,
    "Action, Adventure", "en", "United States of America"
)
```
